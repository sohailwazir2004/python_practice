```python
Caesar Cipher: Implement encode and decode functions. Accept a string and a shift key (integer). 
Encrypt by shifting each letter by key positions (wrap around).
Only shift letters, keep other chars unchanged.
```

```python
s =input("enter the sentence")
shift = int(input("enter the key to shift the text:"))

def encode(s,shift):
    result=""
    for c in s:
        if c.isalpha():
            result +=chr(ord(c)+shift)
        else:
            result+=c
    print(f"the encrypted text is {result}")
    return result

def decode(result, shift):
    original = ""
    for c in result:
        if c.isalpha():
            original += chr(ord(c)-shift)
        else:
            original +=c
    print(f"the decrept text is :{original}")
    
encrypt = encode(s,shift)
decode(encrypt, shift)
```

You're given a list of numbers. Write code to find the maximum difference between any two elements,
where the smaller element must come before the larger one in the list. In other words,
you're looking for the best possible "buy low, sell high" gap — you can't sell before you buy.

```python
prices = [7, 1, 5, 3, 6, 4]
min_price = prices[0]
max_profit = 0
for price in prices:
    if price<min_price:
        min_price = price
    elif price-min_price>max_profit:
        max_profit = price-min_price
print(max_profit)
```

```python
prices = [7, 1, 5, 3, 6, 4]

max_profit = 0
for buy_day in prices:
    for sell_day in range(buy_day+1,len(prices)):
        profit = prices[sell_day]-prices[buy_day]
        if profit>max_profit:
            max_profit=profit
print(max_profit)
        
```

You're given a list of numbers representing daily temperatures. 
Write code to count how many "warmer streaks" exist — meaning, 
count how many days had a strictly higher temperature than the day right before them. 
The very first day doesn't count, since there's nothing before it to compare against.

```python
temps = [30, 32, 31, 35, 35, 38, 36]
# Expected output: 3
# (32>30 ✅, 31>32 ❌, 35>31 ✅, 35>35 ❌ not strictly higher, 38>35 ✅, 36>38 ❌)\
count = 0
for i in range(1,len(temps)):
    if temps[i] > temps[i-1]:
        print(f"{temps[i]} > {temps[i+1]}")
        count+=1
print(count)       
```
