```python
You're given a list of words.
Write code to count how many times each word appears, storing the result in a dictionary.
```

```python
words = ["apple", "banana", "apple", "orange", "banana", "apple"]
dic = {}

for word in words:
    if word in dic:
        dic[word] += 1
    else:
        dic[word] = 1

print(dic)
```

```python
You're given two lists of student names. Write code to find which students are in both lists,
using sets — without writing a manual loop to compare them one by one.
```

```python
You're given a list of tuples, where each tuple holds a student's name and their score.
Write code to convert this into a dictionary, where the name is the key and the score is the value.
```

```python
data = [("Ali", 85), ("Sara", 90), ("Ahmed", 78)]
# Expected output: {'Ali': 85, 'Sara': 90, 'Ahmed': 78}
dic = {}

for d in data:
    dic[d[0]] = d[1]
print(dic)
for name,score in data:
    dic[name] = score 
print(dic)
```

```python
You're given a dictionary of products and their prices.
Write code using dictionary comprehension to create a 
new dictionary containing only the products that cost more than 500.
```

```python
products = {
    'phone': 800,
    'charger': 200,
    'laptop': 1200,
    'earphones': 150,
    'tablet': 600
}
# Expected output: {'phone': 800, 'laptop': 1200, 'tablet': 600}
dic = {}
for key,value in products.items():
   if value> 500:
       dic[key] = value
print(dic)

dic = {key:value for key,value in products.items() if value > 500}
print(dic)
```

```python
You're given a dictionary where each key is a student name and each
value is a list of their scores. Write code to find which students have no duplicate scores in their list.
```

```python
students = {
    'Ali': [85, 90, 85, 78],
    'Sara': [90, 78, 85, 92],
    'Ahmed': [70, 70, 80, 90],
    'Zara': [88, 92, 76, 95]
}
# Expected output: ['Sara', 'Zara']
for key,value in students.items():
    if len(value) == len(set(value)):
        print(key)
```
